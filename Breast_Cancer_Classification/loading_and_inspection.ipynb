## Target Variable

The target variable is `Diagnosis`, which contains two classes:

- `B` = Benign (non-malignant)
- `M` = Malignant

## Important Note

This is a medical classification dataset. Model predictions are intended for educational experimentation and research purposes only, not for clinical diagnosis or medical decision-making.
